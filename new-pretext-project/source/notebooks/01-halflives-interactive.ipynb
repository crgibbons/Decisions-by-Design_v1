## Part 2: The Exponential Decay Model

We model medication concentration $H(t)$ at time $t$ (in days) with the differential equation:

$$\frac{dH}{dt} = -kH$$

where:
- $H(t)$ is the hormone concentration at time $t$
- $k > 0$ is the elimination rate constant
- The negative sign indicates concentration decreases over time

**Interpretation:** The rate at which the body eliminates medication is proportional to how much is present. This is a common model for first-order elimination.

**Solution:** The general solution to this ODE is:

$$H(t) = H_0 e^{-kt}$$

where $H_0$ is the initial concentration.

## Part 3: Half-Life and the Elimination Constant

The **half-life** $T_{1/2}$ is the time for concentration to drop to half its initial value.

Setting $H(T_{1/2}) = \frac{1}{2} H_0$:

$$H_0 e^{-k T_{1/2}} = \frac{1}{2} H_0$$

$$e^{-k T_{1/2}} = \frac{1}{2}$$

$$-k T_{1/2} = \ln(0.5) = -\ln(2)$$

$$k = \frac{\ln(2)}{T_{1/2}}$$

For testosterone with $T_{1/2} = 8$ days:

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint

# Parameters
T_half = 8  # days
k = np.log(2) / T_half

print(f"Half-life: {T_half} days")
print(f"Elimination rate constant k = {k:.4f} per day")
print(f"This means approximately {100*k:.2f}% of the hormone is eliminated per day")
```

## Part 4: Concentration Over Time

Let's visualize how testosterone concentration decays after a single injection.

```python
# Time array (days)
t = np.linspace(0, 40, 200)

# Initial concentration (normalized to 1 for relative comparison)
H0 = 1.0
H = H0 * np.exp(-k * t)

# Create plot
plt.figure(figsize=(10, 6))
plt.plot(t, H, 'b-', linewidth=2, label='Concentration')
plt.axhline(y=0.5, color='r', linestyle='--', alpha=0.7, label='50% (Half-life)')
plt.axvline(x=8, color='r', linestyle='--', alpha=0.7)

# Mark each half-life
for i in range(1, 5):
    t_i = i * T_half
    h_i = H0 * (0.5 ** i)
    plt.plot(t_i, h_i, 'ro', markersize=8)
    plt.text(t_i + 1, h_i + 0.02, f'{100*h_i:.1f}%', fontsize=10)

plt.xlabel('Time (days)', fontsize=12)
plt.ylabel('Hormone Concentration (relative to initial)', fontsize=12)
plt.title('Testosterone Concentration Decay Over Time (T₁/₂ = 8 days)', fontsize=14)
plt.grid(True, alpha=0.3)
plt.legend(fontsize=11)
plt.tight_layout()
plt.show()

print("Key observations:")
for i in range(1, 5):
    t_i = i * T_half
    h_i = 100 * (0.5 ** i)
    print(f"After {t_i} days ({i} half-lives): {h_i:.2f}% remains")
```

## Part 5: Student Exercise 1 – Concentration Remaining

**Question:** A patient stops testosterone treatment with an initial concentration of $H_0$. Calculate the percentage remaining after the following times:

1. After 8 days
2. After 16 days
3. After 24 days
4. After 32 days

**Your turn:** Complete the calculations in the cell below.

```python
# Student exercise: Calculate remaining concentration at different times
times = [8, 16, 24, 32]  # days

print("Percentage of hormone remaining:")
print("-" * 40)

for t_val in times:
    # H(t) = H0 * exp(-k * t)
    # We want the percentage: 100 * H(t) / H0
    percentage = 100 * np.exp(-k * t_val)
    print(f"At t = {t_val:2d} days: {percentage:6.2f}%")

print()
print("Notice the pattern: each 8-day period cuts the concentration in half.")
```

## Part 6: Student Exercise 2 – Interpretation

```python
# Solution to Q1: Time to reach 25%
# H(t) = 0.25 * H0
# exp(-k*t) = 0.25
# -k*t = ln(0.25)
# t = -ln(0.25) / k

t_25 = -np.log(0.25) / k
print(f"Q1: Time to reach 25% concentration: {t_25:.2f} days")
print(f"    (This is 2 half-lives, or 2 × 8 = 16 days)")
print()

# Solution to Q2: Concentration after 2 weeks off
t_interrupt = 14  # days
conc_after_interrupt = 100 * np.exp(-k * t_interrupt)
print(f"Q2: After 2 weeks (14 days) off medication: {conc_after_interrupt:.2f}% remains")
print()

# Q3 is a reflection question
print("Q3: Why this matters for policy:")
print("    - Access interruptions have gradual, not instant effects")
print("    - Policy must account for biological timescales")
print("    - Patients may experience withdrawal symptoms over weeks, not days")
```

## Part 7: Follow-Up Exploration – Dosing Schedules

In practice, patients take repeated doses rather than a single dose. Let's explore what happens when a patient gets weekly injections.

```python
# Repeated dosing: patient gets an injection every 7 days
# Each injection adds D units; between injections, concentration decays

D = 100  # dose amount (arbitrary units)
injection_interval = 7  # days
num_injections = 6

# Simulate the concentration profile
t_detailed = np.linspace(0, num_injections * injection_interval, 1000)
H_repeated = np.zeros_like(t_detailed)

for i in range(num_injections):
    injection_time = i * injection_interval
    # For times after this injection, add the contribution from this dose
    t_since_injection = t_detailed - injection_time
    # Only count if time is after the injection
    mask = t_since_injection >= 0
    H_repeated[mask] += D * np.exp(-k * t_since_injection[mask])

# Plot
plt.figure(figsize=(12, 6))
plt.plot(t_detailed, H_repeated, 'b-', linewidth=2, label='Total concentration')

# Mark injections
for i in range(num_injections):
    injection_time = i * injection_interval
    plt.axvline(x=injection_time, color='r', linestyle='--', alpha=0.5)
    plt.plot(injection_time, H_repeated[np.argmin(np.abs(t_detailed - injection_time))], 'ro', markersize=8)

plt.xlabel('Time (days)', fontsize=12)
plt.ylabel('Total Hormone Concentration (relative units)', fontsize=12)
plt.title('Concentration with Weekly Injections', fontsize=14)
plt.grid(True, alpha=0.3)
plt.legend(fontsize=11)
plt.tight_layout()
plt.show()

print("With repeated dosing:")
print(f"- Each injection adds {D} units")
print(f"- Concentration fluctuates between a minimum and maximum")
print(f"- Eventually reaches a steady-state oscillation")
```
